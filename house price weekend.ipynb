```python
target_column = "Y house price of unit area"
unwanted_columns = ["No", "X1 transaction date"]



data_clean = Data.drop(columns= unwanted_columns)
data_clean
```

```python
features = data_clean.drop(columns=target_column)
target = data_clean[target_column]
```

# Train-test Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test, = train_test_split(features,target,test_size=0.2, random_state=30)
```

```python
len(x_train) == len(y_train) # check if  x_train and y_train have the samev number of rows
```

```python
len(y_test) == len(y_test) # check if y_test and y_test have the same number of rows
```

# Preprocessing

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler
```

```python
scaled_data = scaler.fit_transform(features)  #scales all the feature
scaled_data  # every scaler will have mean 0 standard deviation 1
```

```python
scaled_data[:,0].mean()  # mean of the first column ofv the transformed data
```

```python
scaled_data[:,0].std()  # standard deviation of the first column of the transformed data
```

 # Model bulding

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_regression = LinearRegression()
```

```python
linear_regression.fit(x_train,y_train)
```

```python
linear_regression.coef_   # weight for each feature
```

```python
linear_regression.intercept_  #  intercept
```

#  Train Performance

```python
y_pred_train = linear_regression.predict(x_train) # predict the prices using train data
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse = mean_squared_error(y_train,y_pred_train)
mse
```

```python
import numpy as np
```

```python
np.sqrt(mse)
```
